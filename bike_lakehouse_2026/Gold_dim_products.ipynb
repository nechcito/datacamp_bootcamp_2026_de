#The Transformation Logic

```python
query=("""
SELECT
       ROW_NUMBER() OVER (ORDER BY cp.start_date,cp.product_number) AS product_key,
       cp.product_id,
       cp.product_number,
       cp.product_name,
       cp.category_id,
       pt.category,
       pt.subcategory,
       pt.maintenance_flag,
       cp.product_line,
       cp.start_date
FROM silver.crm_products AS cp
LEFT JOIN silver.erp_product_category AS pt
       ON cp.category_id = pt.category_id
       """)
df=spark.sql(query)
```

#Writing  Gold Table

```python
df.write.mode("overwrite").format("delta").saveAsTable("gold.dim_products")
```
